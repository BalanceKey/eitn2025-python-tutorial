# Data structures in Python

```python
# Variables are created when you assign a value to it
x = 5
y = "I'm using Python in Marseille !"
z = False
f = 31.0
F = 'Panisses' # Note: Python is case-sensitive, so F and f are different variables
Y = ' : recommended by the Python tutor.'

print('Value of x:', x)    # prints the value of x
print('Sum of x and f:', x+f)  # prints the sum of x and f

if not z: 
    print(y) # prints the value of y if z is False

print(F + Y) # concatenates and prints the strings F and Y
```

```python
# Variables can have different data types
a_integer = 10        # Integer
a_float = 10.5        # Float
a_string = "10"       # String
a_boolean = True      # Boolean
a_list = [1, 2, 3]    # List
a_tuple = (1, 2, 3)   # Tuple
a_dict = {'key': 'value'} # Dictionary

# You can check the type of a variable using the type() function
print(type(a_integer))  # Output: <class 'int'>
print(type(a_float))    # Output: <class 'float'>
print(type(a_string))   # Output: <class 'str'>
print(type(a_boolean))  # Output: <class 'bool'>
print(type(a_list))     # Output: <class 'list'>
print(type(a_tuple))    # Output: <class 'tuple'>
print(type(a_dict))     # Output: <class 'dict'>
```

```python
# Lists are ordered, mutable collections of items, defined using square brackets [], and can contain items of different data types.
# List example
a = [2, 3, -1, 4]

print('List a contents:')
print(a)      # Output the entire list
print(type(a))  # Check the type of the list
print('The third element of the list: ', a[2])   # Accessing the third element of the list
print('The length of the list:', len(a)) # Getting the length of the list
```

```python
# Adding an element to the end of the list
a.append('Hola')
print('Updated list:', a)  # Output the updated list
```

```python
# Removing an element from the list
a.remove(-1)
print('List after removal:', a)  # Output the list after removal
```

```python
# Dictionnaries are unordered, mutable collections of key-value pairs, defined using curly braces {}, where each key is unique.
# Dictionary example
b = {"name": "Alice", "age": 30}
print('Dictionary b contents:')
print(b)      # Output the entire dictionary
print(type(b))  # Check the type of the dictionary
print('Dictionary keys:', b.keys())  # Getting all keys in the dictionary
print('Dictionary values:', b.values())  # Getting all values in the dictionary
print(b["name"])  # Accessing the value associated with the key "name"
print(b.get("age"))  # Getting the value associated with the key "age"
```

```python
# Tuples are ordered, immutable collections of items, defined using parentheses (). They also have a smaller memory footprint compared to lists. 
# For example, tuples can be used to represent fixed collections of items, such as coordinates (x, y) or RGB color values (r, g, b).
# Tuple example
c = (1, 2, 3)
print('Tuple c contents:')
print(c)      # Output the entire tuple
print(type(c))  # Check the type of the tuple
print('The first element of the tuple:', c[0])  # Accessing the first element of the tuple
print('The length of the tuple: ', len(c))  # Getting the length of the tuple

# Note: Tuples are immutable, so you cannot add or remove elements after creation.
c[0] = 10  # This will raise a TypeError since tuples do not support item assignment
```

# Loops and if statements
Let's see how python loops and if statements are implemented

```python
for number in range(2, 11, 2):
    print(number)
```

```python
players = [
    "Tilo",
    "Fiona",
    "Ray",
]
for player in players: # iterate through each player in the players list
    print(player)
```

```python
# Using enumerate to get index and player as a tuple
for i, player in enumerate(players):
    print(i, player)
```

```python
# Importantly, Python doesn't care about what data type we are looping over, as long as the object is iterable.
for player in players:
    print(player)
    for letter in player:
        print(letter)
    print('-'*5)
```

```python
# We also have while loops, which continue executing as long as a certain condition is true.
message, count = "end of loop", 1 # we can define multiple variables in one line
while count <= 5:
    print(f"{count}")
    count += 1  # Increment count to avoid infinite loop

print(message)
```

```python
# If-elif-else statements allow you to execute different blocks of code based on certain conditions.
x = 1
if x > 2:
    print("kind of crazy")
elif x == 2:
    print("same")    
elif x == 3:
    print("say what?")
else:
    print("I give up")

print("\n Exercise: Can you make the previous code print 'kind of crazy'?")
```

# Functions in Python

```python
# This is a function 
def my_function():
    print("Hello from a function!") # prints a message

my_function() # calls the function
```

```python
# This is a function with parameters and a return value
def sum_of_squares(ifrom, ito): 
    squares = [i**2 for i in range(ifrom, ito+1)] # List comprehension to generate squares
    return sum(squares)                           # Return the sum of the squares

result = sum_of_squares(1, 10) # Calculate the sum of squares from 1 to 10
print("Sum of squares from 1 to 10 is:", result) 
```

```python
# Functions can also have default parameters
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()          # Uses default parameter
greet("Alice")   # Overrides default parameter
greet(F)         # Uses variable F as parameter
```
