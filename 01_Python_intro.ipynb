# Data structures in Python

```python
# Variables are created when you assign a value to it
x = 5
y = "I'm using Python in Marseille !"
z = False
f = 31
F = 'Panisses' # Note: Python is case-sensitive, so F and f are different variables
Y = ' : recommended by the Python tutor.'

print(x)    # prints the value of x
print(x+f)  # prints the sum of x and f

if not z: 
    print(y) # prints the value of y if z is False

print(F + Y) # concatenates and prints the strings F and Y
```

```python
# Variables can have different data types
a_integer = 10        # Integer
a_float = 10.5        # Float
a_string = "10"       # String
a_boolean = True      # Boolean
a_list = [1, 2, 3]    # List
a_tuple = (1, 2, 3)   # Tuple
a_dict = {'key': 'value'} # Dictionary
# You can check the type of a variable using the type() function
print(type(a_integer))  # Output: <class 'int'>
print(type(a_float))    # Output: <class 'float'>
print(type(a_string))   # Output: <class 'str'>
print(type(a_boolean))  # Output: <class 'bool'>
print(type(a_list))     # Output: <class 'list'>
print(type(a_tuple))    # Output: <class 'tuple'>
print(type(a_dict))     # Output: <class 'dict'>
```

```python
# List example
a = [2, 3, -1, 4]

print(a)      # Output the entire list
print(a[2])   # Accessing the third element of the list
print(len(a)) # Getting the length of the list
```

```python
# Adding an element to the end of the list
a.append(5)
print('Updated list:', a)  # Output the updated list
```

```python
# Removing an element from the list
a.remove(-1)
print('List after removal:', a)  # Output the list after removal
```

```python
# Dictionary example
b = {"name": "Alice", "age": 30}
print(b["name"])  # Accessing the value associated with the key "name"
print(b.get("age"))  # Getting the value associated with the key "age"
```

```python
# Tuple example
c = (1, 2, 3)
print(c[0])  # Accessing the first element of the tuple
print(len(c))  # Getting the length of the tuple
```

# Functions in Python

```python
# This is a function 
def my_function():
    print("Hello from a function!") # prints a message

my_function() # calls the function
```

```python
# This is a function with parameters and a return value
def sum_of_squares(ifrom, ito): 
    squares = [i**2 for i in range(ifrom, ito+1)] # List comprehension to generate squares
    return sum(squares)                           # Return the sum of the squares

result = sum_of_squares(1, 10) # Calculate the sum of squares from 1 to 10
print("Sum of squares from 1 to 10 is:", result) 
```

```python
# Functions can also have default parameters
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()          # Uses default parameter
greet("Alice")   # Overrides default parameter
greet(F)         # Uses variable F as parameter
```
